```python
import cv2
import uuid # unique identifier
import os
import time
```

```python
# Define images to collect
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'k', 'l','m', 'n','o','p', 'q', 'r', 's', 't','u','v','w', 'x', 'y']
number_imgs = 5

# setup_folders
IMAGES_PATH = "image_folders"

for label in labels:
    path = os.path.join(IMAGES_PATH, label)
    if not os.path.exists(path):
        os.makedirs(path, exist_ok=True)
```

```python
for label in labels:
    label_path = os.path.join(IMAGES_PATH, label)
    os.makedirs(label_path, exist_ok=True)

    cap = cv2.VideoCapture(0)
    print(f'[INFO] Collecting images for letter "{label}"')
    print('Pess "c" to capture the image, "q" for quit.')

    imgnum = 0

    while imgnum < number_imgs:
        ret, frame = cap.read()
        cv2.imshow('Press "c" to capture', frame)

        key = cv2.waitKey(1)

        if key & 0xFF == ord('c'):
            imgname = os.path.join(label_path, f'{label}.{str(uuid.uuid1())}.jpg')
            cv2.imwrite(imgname, frame)
            print(f'[INFO] Image {imgnum+1}/{number_imgs} saved: {imgname}')
            imgnum += 1
            time.sleep(0.5)

        elif key & 0xFF == ord('q'):
            print('[INFO] Exiting.')
            break

    cap.release()
    cv2.destroyAllWindows()
```
